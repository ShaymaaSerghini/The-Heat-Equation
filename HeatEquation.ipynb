The heat equation we will solve is:

$ \frac{\partial T}{\partial t} = \alpha \frac{\partial T^2}{\partial x^2} - \beta (T^4 - S^4) = 0, \quad (1)$

Here $T(x,t)$ is the time dependent temperature profile along the 1d bar that we want to solve for and $S(x,t)$ is a function that represents the temperature of the surroundings, which provides a source or sink of heat if its value is different to the temperature of our object.

To solve the problem we will need to specify the function $S(x,t)$. This will have the form:

$ S(x,t) = A + B \cos(\omega t) e^{-(x/C)^2}  \quad (2)$ 

We will also need the initial Temperature profile:

$ T(x,t=0) = T_0 (1 + e^{-x^2})  \quad (3) $ 

And we will assume that the boundary conditions are such that 

$ T(x=-L/2) = T(x=L/2) = T_0 $

We need to make a class called `Surroundings`, this class has methods to:

    1- Calculate the spatial profile S(x) at a given time t of the equation S(x,t).
    2- Calculate the temporal profile S(t) at a given spatial coordinate x.
    
S(x,t) is the temperature of the surroundings at position x and time t. it acts as a source of heat for the bar.

The parameters S(x,t) are:

**A** is a constant , and it is the baseline temperature, ie the temperature when no oscillations or other effects are considered.

**B** is a constant representing the amplitude of oscillation of the heat equation, it controls how much the temperature varies or oscillates around the baseline value A.

**C** is the spatial decay factor, it controls how the effect diminishes with distance along the bar.

**Omega** $\omega$ controls how quickly the temperature oscillates over time. A large $\alpha$ means faster heat diffusion.

**Beta** $\beta$ controls the rate of heat loss or gain due to the difference between the temperature of the bar and the surroundings. A small $\beta$ implies slower heat exchange. 

**L** is the length of the bar, Sets the spatial domain of the problem, with $x = [-L/2, L/2]$

$ S(x,t) = A + B \cos(\omega t) e^{-(x/C)^2}  \quad$

The exponential term in S(x,t) ensures that the temperature profile decays as x moves away from the centre (x=0), and C controls how sharply this decay happens, smaller C means faster decay.

The cosine term adds a periodic oscillation in time, making the surroundings' temperature vary, and omega determines the frequency of oscillation.

```python
#Surroundings class for S(x,t)
class Surroundings:
    """
    A class to implement the surroundings temperature profile given by Equation (2).
    
    Methods:
    -spatial_profile(x,t) : returns S(x,t) for a given spatial coordinate x and time t
    -temporal_profile(x,t_values) : returns S(t) for a fixed spatial coordinate x over a range of times t
   
   Attributes:
    - A: Baseline temperature : float.
    - B: Amplitude of oscillation : float.
    - C: Spatial decay length : float.
    - omega: Temporal frequency of oscillation : float.
    """
    #Constructor function
    def __init__(self, A=20.0, B=1.0, C=1.5, omega=0.5):
        
        assert A >= 0, "Baseline temperature A must be positive"
        assert B >= 0, "Amplitude of oscillation B must be positive."
        assert omega > 0, "Frequency ω must be positive."
        assert C > 0, "Spatial decay parameter C must be positive."

        self.A = A  # Baseline temperature
        self.B = B  # Amplitude of oscillation
        self.C = C  # Spatial decay length
        self.omega = omega  # Temporal frequency

    def get_spatial_profile(self, x, t):
        
        """
        Compute the spatial profile S(x) for a given time t.
        
        Attributes:
            x (array-like): Spatial positions (array or list).
            t (float): Time.
        Returns:
        - (array-like): The spatial profile S(x, t) at each position x for the given time t.
            
        """

        assert t >= 0, "Time t must be non-negative."
        return self.A + self.B * np.cos(self.omega * t) * np.exp(-(x / self.C)**2)

    def get_temporal_profile(self, x, t_values):
        
        """
        Compute the temporal profile S(t) for a given spatial coordinate x.
        
        Attributes:
            x (float): Spatial position.
            t_values (array-like): Time values.
        Returns:
        - (list): The temporal profile S(t) at each time in t_values for the given spatial position x.
        
        """
        
            
        assert np.all(t >= 0), "Time t must be non-negative."
        # Compute the temporal profile by evaluating S(x, t) for each time t
        return [self.get_spatial_profile(x, t) for t in t_values]

# Create the Surroundings object
surroundings = Surroundings()
```

#### Why A > 0 and B > 0 ? 

A is the baseline temperature of the surroundings. It represents the temperature when no oscillations or changes are happening. In real life, the surroundings are rarely at zero temperature, there’s usually some baseline warmth. For example, a room might always have a steady temperature of 21°C, this is the logic behind me defining A to be striclty positive.

B is the strength of temperature oscillations. It controls how much the surroundings' temperature changes over time, temperature varies during the day, so it would make sense for B to be strictly positive, otherwise we will be assuming that there are no temperature changes and the surroundings always stay at the baseline temperature if we let B be zero.

**Temperature profile** :

Temperature profile T(x,t) : describes the temperature distribution in the bar, influenced by heat conduction and external heat sources.

The temperature profile T(x,t) changes over time as heat spreads along the bar, and it is governed by the heat equation:

$ \frac{\partial T}{\partial t} = \alpha \frac{\partial T^2}{\partial x^2} - \beta (T^4 - S^4) = 0, \quad $

Unlike the surrounding temperature profile S(x,t), the temperature profile T(x,t) will evolve based on the interaction between heat conduction and the heat exchange between the bar and the surroundings, which causes the temperature to adjust and spread across the bar.

```python
# Constants
A = 20.0
B = 1.0
C = 1.5
omega = 0.5
T0 = 20
alpha = 0.5
beta = 0.0001
L = 10
Nx = 100  # Number of spatial points
Nt = 500  # Number of time steps
dx = L / (Nx - 1)  # Spatial grid size
dt = 0.01  # Time step size
x = np.linspace(-L/2, L/2, Nx)  # Spatial grid
t = np.linspace(0, Nt*dt, Nt)  # Time grid

# Initial temperature profile
T = T0 * (1 + np.exp(-x**2))

#storing temperature at each time step
T_all = np.zeros((Nt, Nx))
T_all[0, :] = T

# Function to compute surrounding temperature S(x,t)
def S(x, t):
    return A + B * np.cos(omega * t) * np.exp(-(x / C)**2)

for n in range(0, Nt-1):
    for i in range(1, Nx-1): 
        S_val = S(x[i], t[n])  # Surrounding temperature
        T_new = T_all[n, i] + dt * (alpha * (T_all[n, i+1] - 2*T_all[n, i] + T_all[n, i-1]) / dx**2 - beta * (T_all[n, i]**4 - S_val**4))
        T_all[n+1, i] = T_new
    
    # Apply boundary conditions
    T_all[n+1, 0] = T0  # Left boundary
    T_all[n+1, -1] = T0  # Right boundary

# Plot the temperature profiles at different time steps
plt.figure(figsize=(10, 6))
for n in range(0, Nt, int(Nt/5)):
    plt.plot(x, T_all[n, :], label=f"t = {t[n]:.2f}")
plt.xlabel("x")
plt.ylabel("Temperature T(x,t)")
plt.title("Temperature Profiles at Different Time Steps")
plt.legend()
plt.grid(True)
plt.show()
```

#### Interpretation of the plot:

- The x-axis represents the position x along the 1D bar.
- The y-axis represents the temperature T(x,t) at each spatial position x.
- The different colored curves represent the temperature profile at different time steps t, with each curve showing how the temperature changes across the bar over time.

This plot shows how the temperature of the bar evolves over time, starting from an initial peak at the centre (the blue line) and gradually smoothing out due to the diffusion process. The surrounding temperature influences the bar's temperature but becomes less significant over time as the bar's temperature distribution spreads out and reaches a steady state.

#### Convergence testing for the solution of the heat equation:

We aim to solve the heat equation using a numerical method RK4 and then test how well the solution converges as we refine the time step dt.

For each time resolution, the final temperature profile is computed. The error is then calculated by comparing the results obtained with different time steps. The key point is to see how the error decreases as the time step dt is halved.

As the time step dt is reduced, the error between solutions for different resolutions should decrease. For a 4th-order method like RK4, the error should decrease by a factor of 16 when dt is halved.

To do this we refer to the convergence test for RK4 in week4Lab_exercises of this module "Topics in Scientific Computing".
Below is the relevant code I will be using:

```python
# Function to calculate the temperature profile using the RK4 method
def rk4_temperature(dt, Nt, Nx, T0, dx, alpha, beta):
    T_all = np.zeros((Nt, Nx))
    T = T0 * (1 + np.exp(-x**2))  # Initial condition
    T_all[0, :] = T

    for n in range(0, Nt-1):
        k1 = np.zeros(Nx)
        k2 = np.zeros(Nx)
        k3 = np.zeros(Nx)
        k4 = np.zeros(Nx)
        
        for i in range(1, Nx-1):  # Loop over spatial points
            
            # Compute k1 : is the initial slope based on the current state.
            S_val = S(x[i], n * dt)  # Surrounding temperature at time step n
            k1[i] = dt * (alpha * (T[i+1] - 2*T[i] + T[i-1]) / dx**2 - beta * (T[i]**4 - S_val**4))

            # Compute k2 : is the slope at the midpoint, using k1 to estimate the state halfway through the time step. 
            S_val = S(x[i], (n + 0.5) * dt)
            k2[i] = dt * (alpha * (T[i+1] + 0.5*k1[i] - 2*T[i] + T[i-1] + 0.5*k1[i-1]) / dx**2 - beta * ((T[i] + 0.5*k1[i])**4 - S_val**4))

            # Compute k3 : is another estimate of the slope at the midpoint using k2.
            S_val = S(x[i], (n + 0.5) * dt)
            k3[i] = dt * (alpha * (T[i+1] + 0.5*k2[i] - 2*T[i] + T[i-1] + 0.5*k2[i-1]) / dx**2 - beta * ((T[i] + 0.5*k2[i])**4 - S_val**4))

            # Compute k4 : is the final slope at the end of the time step, using to estimate the future state.
            S_val = S(x[i], (n + 1) * dt)
            k4[i] = dt * (alpha * (T[i+1] + k3[i] - 2*T[i] + T[i-1] + k3[i-1]) / dx**2 - beta * ((T[i] + k3[i])**4 - S_val**4))

        # Clip the temperature values to prevent overflow
        T = np.clip(T, -1e10, 1e10)  

        # Update temperature for next time step
        T += (k1 + 2*k2 + 2*k3 + k4) / 6
        T_all[n+1, :] = T

    return T_all

# Step sizes for the convergence test 
N_steps_vector = [2048, 4096, 8192]  # Different resolutions 

# Here I had to use finer resolutions, I started with [16 ,32, 64] but RK4 was not converging as expected. I was not able to see the 1/16 error 
#reduction with coarser resolutions.

# Expected error reduction for 4th-order method
order = 16  

# Store old solution for comparison
old_solution = None

# Perform convergence test for different time steps
for isteps, N_steps in enumerate(N_steps_vector):
    dt = 0.0001 / N_steps  
    Nt = 600  # Number of time steps 

    # Solve temperature profile for current resolution using RK4
    T_all_current = rk4_temperature(dt, Nt, Nx, T0, dx, alpha, beta)

    # Extract the final temperature profile at the last time step for comparison
    temp_at_t = T_all_current[-1, :]  # Extract the final time step (last row)

    if isteps > 0:
        # Compare current solution with previous solution
        diff = temp_at_t - old_solution  # Compare last solutions for current and previous dt

        # Plot the error between consecutive solutions
        plt.plot(x, diff, label=f"Error between h={dt} and h={dt*2} at final time step")
        
        # Plot expected error reduction for convergence test
        if isteps == 1:
            plt.plot(x, diff / order, '--', label="Expected error reduction by factor of 16")

    # Store current solution as the old solution for the next iteration
    old_solution = temp_at_t

# Plot configuration
plt.grid()
plt.legend(loc='upper left', bbox_to_anchor=(1, 1), fontsize='small') 
plt.ylabel("Solution error (Temperature Profile)")
plt.xlabel("Position x") 
plt.title("Convergence Test for Temperature Profile using RK4 Method")
plt.show()
plt.tight_layout()
```

#### Interpretation :

For a 4th-order method RK4, we expect the error to decrease by a factor of 16 when halving the time step size. The error between consecutive step sizes (blue and green curves) reduces significantly as h is halved.

The orange dashed line represents the expected behavior for the convergence of the RK4 method, which is a 4th-order method. This means that the error should reduce by approximately a factor of 16 as the time step is halved.

The RK4 is a fourth-order accuracy method, the error of the ratio is 1/16. Thus we are in the convergence regime.

Referring to the lecture notes in Week 8 to check how the polynomial is constructed:

The polynomial is constructed such that it agrees with $f(x)$ at specific points:
   $(x_0, f(x_0)), (x_1, f(x_1)), \dots, (x_n, f(x_n))$
   
   
First, we construct the basis polynomials $L_k(x)$, which are defined to satisfy:


   The formula for $L_k(x)$ is:
   
   $L_k(x) = \prod_{\substack{i=0 \\ i \neq k}}^{n} \frac{(x - x_i)}{(x_k - x_i)}$

The Lagrange interpolant $P_n(x)$ is the weighted sum of these basis functions:

   $P_n(x) = \sum_{k=0}^{n} L_k(x) f(x_k)$

   - The **numerator** of $L_k(x)$: Ensures the polynomial is zero at all $x_i$ except $x_k$.
   - The **denominator**: Normalises the function so that $L_k(x_k) = 1$.

#### What is the Chebyshev Method?

The Chebyshev method uses Chebyshev polynomials, a special type of mathematical function, to solve different numerical problems. It can be used in:

    - Integration 
    - Interpolation
    - Differentiation.
    
#### Why Use Chebyshev Nodes for Interpolation ?


<u> Reduces Runge’s Phenomenon:

Interpolation with equally spaced points leads to oscillations near the boundaries. Chebyshev nodes avoid this by clustering points closer to the boundaries.

Gauss–Lobatto nodes typically reduce oscillations near boundaries.
    
<u> Minimises Errors: 

Chebyshev nodes spread errors more evenly and reduce the maximum error in polynomial interpolation.


<u> Faster Convergence:

Chebyshev nodes allow the interpolation polynomial to approximate the true function more efficiently, especially for smooth functions.  


#### Why are Gauss-Lobato nodes effective:

Gauss-Lobato nodes are the extrema of Chebyshev polynomials. They minimise the interpolation error by concentrating points near the edges of the interval, where Runge's phenomenon occurs. This distribution ensures a more uniform error across the interpolation domain.

The maximum error is computed as the difference between the Lagrange polynomial fit and the true curve at all points. The method iteratively increases the number of nodes until the maximum error is below 0.01, this will show the convergence behaviour of interpolation with Gauss-Lobato nodes.

While Gauss-Lobato points reduce interpolation error and mitigate Runge's phenomenon, they require calculating node positions using trigonometric functions. The computational cost is slightly higher compared to equally spaced points. However, the reduction in interpolation error often justifies the increased complexity.
Generally, Chebyshev methods improve numerical accuracy and efficiency.


```python
# Function to calculate Gauss-Lobato nodes
def compute_gauss_lobato_nodes(n_points, interval):
    a, b = interval
    # Standard Gauss-Lobato nodes in [-1, 1]
    nodes = np.cos(np.pi * np.linspace(0, n_points - 1, n_points) / (n_points - 1))
    # Transform to interval [a, b]
    transformed_nodes = 0.5 * (a + b) + 0.5 * (b - a) * nodes
    return transformed_nodes

# Fit and plot for Gauss-Lobato nodes
def compute_lagrange_fit_gauss_lobato(n_points, label):
    # Calculate Gauss-Lobato nodes in the interval [-L/2, L/2]
    x_data = compute_gauss_lobato_nodes(n_points, [-L/2, L/2])
    T_data = surroundings.get_spatial_profile(x_data, t)

    # Fit the Lagrange polynomial
    lagrange_poly = lagrange(x_data, T_data)

    # Evaluate the polynomial and the true curve
    x_values = np.linspace(-L/2, L/2, 200)  # Fine grid for plotting
    fitted_values = np.polyval(lagrange_poly, x_values)
    true_values = surroundings.get_spatial_profile(x_values, t)

    # Plot the results
    plt.figure(figsize=(10, 6))
    plt.plot(x_values, true_values, label="True Curve S(x, t=2)", color="blue")
    plt.plot(x_values, fitted_values, '--', label=f"Lagrange Fit ({label} points, Gauss-Lobato)", color="orange")
    plt.scatter(x_data, T_data, color="red", label=f"Data Points ({label} points, Gauss-Lobato)")
    plt.title(f"Lagrange Polynomial Fit with {label} Gauss-Lobato Points")
    plt.xlabel("x")
    plt.ylabel("S(x, t=2)")
    plt.legend()
    plt.grid()
    plt.show()

    # Compute the error
    max_error = np.max(np.abs(fitted_values - true_values))
    print(f"Max error with {label} Gauss-Lobato points: {max_error:.5f}")
    return max_error

# Gauss-Lobato fit for 6 points
max_error_6_gauss = compute_lagrange_fit_gauss_lobato(6, "6")

# Gauss-Lobato fit for 9 points
max_error_9_gauss = compute_lagrange_fit_gauss_lobato(9, "9")
```

#### What is the Trapezium Rule:

The Trapezium Rule is a numerical integration method that approximates the area under a curve by dividing the interval into smaller subintervals and approximating the function as a straight line over each subinterval.

Both methods (python function+ implemented version) are implementing the same underlying numerical integration technique -> the **Trapezium Rule**, hence the outputs should be the same.

#### <u> Advantages: <u>
    
- The Trapezium rule is straightforward and easy to implement.
    
- It works well for functions that are smooth and approximately linear over small intervals.
   
- Compared to some other methods (e.g., rectangular rule), the Trapezium rule gives better accuracy for a small number of subintervals.

#### <u> Disadvantages: <u>
    
- It approximates the function as linear, so it may not perform well for functions with high curvature or rapid changes.
    
- Slow Convergence.
    
- Not Suitable for Discontinuous Functions.
    
This approximation always **overestimates** or **underestimates** the integral, depending on the function's curvature within the interval.

#### How the Trapezium Rule Works: ( Using Week 8 notes)

-  Split the interval $([a, b]$) into n subintervals of equal width:
    
   $ h = \frac{b - a}{n} $.

- Approximate the function $(f(x)$) as a straight line between each pair of adjacent points. 
    
    $\int_a^b f(x) dx \approx \frac{h}{2} \left[ f(x_0) + 2f(x_1) + 2f(x_2) + \dots + 2f(x_{n-1}) + f(x_n) \right]$
   
    Where:
    $ (x_0 = a, x_1 = a + h, x_2 = a + 2h, \dots, x_n = b) $ 

    
I aim for fairly close values since both the `scipy.integrate.trapezoid` and the method I will implement use equivalent formulas, meaning that the computed areas under the curve are the same. And this is pretty normal because when both methods are given the same input, they will produce the same output : the estimated integral.

```python
#Estimating the integral:
t = 1  # time at which the integral is evaluated
L = 10  # length of the bar
N = 11  # number of equally spaced points
x_values = np.linspace(-L / 2, L / 2, N)  # equally spaced points
S_values = surroundings.get_spatial_profile(x_values, t)  # S(x, t) at t=1
S_shifted = S_values + 273  # Shifted by 273 as per the formula

# Trapezium rule implementation
def compute_trapezium_rule(x, y):
    """
    Implements the trapezium rule from scratch.
    x_array: sorted array of x
    y_array: function values at these x
    Returns the integral approximation.
    
    """
    
    assert len(x) == len(y), "x and y must have the same length."
    assert len(x) > 1, "There must be at least two points to apply the trapezium rule."
    
    h = x[1] - x[0]
    integral = h * (0.5 * y[0] + np.sum(y[1:-1]) + 0.5 * y[-1])
    return integral

# Calculate the integral using the trapezium rule
E_trapezium = compute_trapezium_rule(x_values, S_shifted)

# Verify using scipy.integrate.trapezoid
E_scipy = trapezoid(S_shifted, x_values)

print(f"Estimated integral using Trapezium Rule (implemented): {E_trapezium}")
print(f"Estimated integral using scipy.integrate.trapezoid: {E_scipy}")

E_trapezium, E_scipy
```

#### Convergence Test Explanation:

In this convergence test, we aim to verify the second-order convergence of the trapezium rule method used to integrate the temperature profile. We compare the results at three different resolutions to see if the error scales as expected for a second-order method.

h = \frac{x_{\text{end}} - x_{\text{start}}}{N}
\end{equation}
 
The trapezium rule is a second-order accurate method, meaning that if we halve the step size, the error should reduce by approximately a factor of 4.

The convergence condition we are testing is:

\begin{equation}
\frac{\lVert E_{N=64} - E_{N=32} \rVert}{\lVert E_{N=32} - E_{N=16} \rVert} \approx \frac{1}{4}
\end{equation}
 
If the error reduction between successive resolutions has a scaling factor of approximately 1/4, then we can conclude that the trapezium method is converging with the expected second-order accuracy, and thus we are "in the convergence regime".

Here is what we could do to prove that the trapezium rule is within the converegenve regime:

Step 1 :

Define 3 different step sizes and since we do not know the solutions we use 3 resolutions to test convergence .

Step 2 :

We find the high-resolution (HR) difference, defined as E3-E2, and the Low-Resolution (LR) difference, defined as E2-E1.

Where E3 defines the High resolution (HR) ,E2 defines the medium-resolution (MR), and E1 defines the low-resolution (LR).

Step 3 :

The differences LR and HR are plotted, we are expecting the plot to show that the solution converges , as the step size decreases the difference between two consecutive solutions also decreases, and the differences should scale by a factor of 4.

So we follow the same steps as above but in this case we find the true integral using N= 1000 points, for each resolution, the absolute error is calculated as the difference between the computed integral and the true integral:

#### Background on the Pseudospectral Method:

The pseudospectral method (PS) is a numerical technique used to solve mathematical problems, like finding derivatives or solving equations, with high accuracy.

For example : 

Say you have a wavy curve, like a temperature profile along a metal rod. You want to understand how it changes, like its slope (first derivative) or how fast the slope changes (second derivative). 

Normally, if I had no idea what the PS method is I would divide the curve into small chunks, connect those points with straight lines, and approximate the changes. It's like drawing a "connect-the-dots" version of the curve, which is simple but not super accurate.

The PS method instead of connecting dots, it uses smooth, wavy mathematical functions (like sines and cosines, or polynomials) that fit the entire curve. These functions naturally match the shape of the curve much better.

#### How to pick the points:

Instead of choosing points equally spaced (like dividing the rod into segments of equal size), the pseudospectral method uses special, smartly chosen points the Gauss-Lobato points where the math works better.

Gauss-Lobato points are special points used in numerical methods, to make calculations like integration more accurate. These points are distributed in a very clever way across an interval, often denser near the edges and sparser in the middle. This helps to handle boundary effects and reduce numerical errors.

These points are spread more densely near the edges of the curve, where tricky changes often happen, and more sparsely in the middle.( You can see this in q1d "Lagrange Fit with 6 Equally Spaced Points" plot and in q1e Lagrange Fit with 6 or 9 Gauss Lobato Nodes ). This reduces numerical oscillations.

#### What is the Stencil Method?

A stencil is like a recipe or a small formula that uses values at a few nearby points to compute something like a slope (first derivative) or curvature (second derivative).
It's like saying, "To estimate the slope at this point, I’ll use this point, the one to the left, and the one to the right, and combine them in a specific way."

We think of the stencil method as a way to approximate the slope or curvature of a curve (like a hill or a roller coaster) by looking at nearby points instead of trying to figure out everything at once. It’s like drawing a smooth curve by connecting a few dots instead of drawing the entire thing freehand.

When we want to find how steep something is at a point : the first derivative or how quickly that steepness changes : the second derivative , we can use nearby points to approximate it. The stencil method gives us a formula that tells us how to combine these points to get a good guess for the slope or curvature.

So to approximate the slope at a point, the stencil uses the values of the function (temperature, in our case) at points close to it.

For a fourth-order accurate stencil, we use five points:

- Two points to the left $ (x−2h , x−h)$,

- The point itself (x),

- Two points to the right $ (x+h , x+2h)$.


Using the online calculator [finite difference calculator](https://web.media.mit.edu/~crtaylor/calculator.html) we obtain the following:

_For the first derivative:_ 

To achieve a fourth-order accurate stencil, five points are needed:

$$
f'(x_0) \approx \frac{-f(x - 2h) + 8f(x - h) - 8f(x + h) + f(x + 2h)}{12h}
$$

Here, h is the distance between the points (grid spacing). The numbers (-1, 8, -8, 1) are the weights, which are carefully calculated to make the approximation as accurate as possible.

_For the second derivative:_ 

To approximate how quickly the slope changes, a second-order accurate stencil uses only three points:

- One point to the left $ (x−h)$,
- The point itself (x),
- One point to the right $ (x+h)$.


The 4th order accurate stencil for the second derivative requires three points:

$$
f''(x_0) \approx \frac{f(x - h) - 2f(x) + f(x + h)}{h^2}
$$

The numbers (1, -2, 1) are again the weights chosen to ensure the approximation is as good as possible.

These weights can be derived using Lagrange polynomials. These polynomials are used to interpolate or guess the value of a function at a point based on nearby points. The stencil coefficients are carefully chosen to minimise error when approximating the derivative.

So we aim to create matrices, or a table of numbers that encode the stencil formulas for every point in the grid.

Each row of the matrix corresponds to a point, and the weights from the stencil are used to fill the matrix.

When we multiply the matrix by the values of the function at the grid points, it gives us the approximate derivatives (slopes and curvatures) at each point.

#### Why do we need to use stencils?

We’re studying how temperature changes along a rod or in space.

To predict these changes, we need to calculate derivatives of the temperature function.

Since we don’t know the exact function, we use a grid of points and approximate the derivatives using stencils.

At the boundaries, where the full stencil is unavailable, one-sided derivatives are used. These are lower-order accurate (third-order for the first derivative, second-order for the second derivative) but sufficient for the edges.

#### Multiple Frequencies:

If the system contains multiple frequencies, finer/ higher spatial and temporal resolutions are required to capture high-frequency components, imposing stricter CFL conditions.

#### RK2 Stability:

While RK2 is more accurate than forward Euler  as it is second-order in time, it still requires adherence to the CFL condition for stability.


```python
class HeatEquationFD:
    """
    Solves the heat equation using finite differences and explicit time integration (RK2).
    
    Attributes: 
    - spatial_grid : The spatial grid for the solution.
    - initial_temperature : Initial temperature profile.
    - max_time : Maximum simulation time.
    - diffusivity : Thermal diffusivity (α).
    - frequency : Frequency of the oscillating surroundings source.
    
    Methods:
    - __init__: Initialise the heat equation solver with given parameters.
    - get_time_derivative: Compute the time derivative of the temperature profile.
    - integrate_using_rk2: Integrate the heat equation using the RK2 method.
    """
    def __init__(self, spatial_grid, initial_temperature, max_time, diffusivity=0.5, frequency=0.5):
        assert len(spatial_grid) == len(initial_temperature), "Spatial grid and initial temperature must have the same length."
        assert max_time > 0, "Maximum time must be positive."
        assert diffusivity > 0, "Diffusivity must be positive."
        assert frequency > 0, "Frequency must be positive."

        self.diffusivity = diffusivity
        self.frequency = frequency
        self.max_time = max_time
        self.spatial_grid = spatial_grid
        self.grid_spacing = spatial_grid[1] - spatial_grid[0]
        self.temperature = initial_temperature
        self.num_grid_points = len(spatial_grid)
        
        # Second derivative matrix (2nd order accurate stencil)
        self.second_derivative_matrix = get_FD_second_derivative_matrix(self.num_grid_points, self.grid_spacing)

    def get_time_derivative(self, current_time, temperature_profile):
        """
        Compute the time derivative of the temperature profile using the heat equation.
        """
        # Compute surroundings source S(x, t)
        surroundings = 20 + np.cos(self.frequency * current_time) * np.exp(-(self.spatial_grid / 1.5) ** 2)
        
        # Heat equation:
        return self.diffusivity * np.dot(self.second_derivative_matrix, temperature_profile) + (surroundings - temperature_profile)

    def integrate_using_rk2(self, time_step, num_steps):
        """
        Integrate the heat equation using the explicit RK2 method.
        """
        # CFL condition check
        cfl_limit = self.grid_spacing ** 2 / (2 * self.diffusivity)
        assert time_step <= cfl_limit, f"Time step too large! CFL condition requires Δt ≤ {cfl_limit:.5e}"

        # Initialise variables
        temperature_profile = self.temperature
        time_points = [0]
        solutions = [temperature_profile.copy()]

        # Time integration loop
        for step in range(num_steps):
            current_time = step * time_step
            
            # RK2 (midpoint method)
            k1 = time_step * self.get_time_derivative(current_time, temperature_profile)
            k2 = time_step * self.get_time_derivative(current_time + time_step / 2, temperature_profile + k1 / 2)
            temperature_profile += k2
            
            # Store the solution
            time_points.append(current_time + time_step)
            solutions.append(temperature_profile.copy())

        self.temperature = temperature_profile
        return np.array(time_points), np.array(solutions)

def get_FD_second_derivative_matrix(N, h):
    """
    Constructs a second derivative finite difference matrix with 2nd order accuracy.
    
    Parameters:
    - N (int): Number of grid points.
    - h (float): Grid spacing.
    
    Returns:
    - np.ndarray: Second derivative matrix of size N x N.
    """
    one_over_h2 = 1.0 / (h * h)
    D2 = np.zeros((N, N))

    for i in range(N):
        if i == 0:  # First boundary point
            D2[i, i] = 1.0 * one_over_h2
            D2[i, i + 1] = -2.0 * one_over_h2
            D2[i, i + 2] = 1.0 * one_over_h2
        elif i == N - 1:  # Last boundary point
            D2[i, i] = 1.0 * one_over_h2
            D2[i, i - 1] = -2.0 * one_over_h2
            D2[i, i - 2] = 1.0 * one_over_h2
        else:  # Interior points
            D2[i, i] = -2.0 * one_over_h2
            D2[i, i - 1] = 1.0 * one_over_h2
            D2[i, i + 1] = 1.0 * one_over_h2

    return D2

def get_FD_first_derivative_matrix(N, h):
    """
    Constructs a first derivative finite difference matrix with 4th order accuracy.
    
    Parameters:
    - N (int): Number of grid points.
    - h (float): Grid spacing.
    
    Returns:
    - np.ndarray: First derivative matrix of size N x N.
    """
    one_over_h = 1.0 / h
    D = np.zeros((N, N))

    for i in range(N):
        if i >= 2 and i < N - 2:  # Interior points
            D[i, i - 2] = 1.0 / (12.0 * one_over_h)
            D[i, i - 1] = -8.0 / (12.0 * one_over_h)
            D[i, i + 1] = 8.0 / (12.0 * one_over_h)
            D[i, i + 2] = -1.0 / (12.0 * one_over_h)
        elif i == 0:  # First boundary point
            D[i, i] = -3.0 * one_over_h
            D[i, i + 1] = 4.0 * one_over_h
            D[i, i + 2] = -1.0 * one_over_h
        elif i == N - 1:  # Last boundary point
            D[i, i] = 3.0 * one_over_h
            D[i, i - 1] = -4.0 * one_over_h
            D[i, i - 2] = 1.0 * one_over_h

    return D

# Define parameters and initial conditions
domain_length = 10
num_grid_points = 50
spatial_grid = np.linspace(-domain_length / 2, domain_length / 2, num_grid_points)
initial_temperature = 20 * (1 + np.exp(-spatial_grid ** 2))
maximum_time = 1.0
time_step = 0.01
number_of_steps = int(maximum_time / time_step)

# Solve the heat equation using RK2
heat_equation = HeatEquationFD(spatial_grid, initial_temperature, maximum_time)
time_points, solutions = heat_equation.integrate_using_rk2(time_step, number_of_steps)

# Plot the solution at different time steps
plt.figure(figsize=(10, 6))
for i, current_time in enumerate(time_points):
    if i % 10 == 0:  # Plot every 10th time step
        plt.plot(spatial_grid, solutions[i], label=f't={current_time:.2f}')

plt.xlabel('x (spatial coordinate)')
plt.ylabel('T(x, t) (temperature profile)')
plt.title('Heat Equation Solved Using RK2 and Finite Differences')
plt.legend()
plt.xlim(-4, 4)
plt.ylim(20, 45)
plt.grid()
plt.show()
```

#### What is the Pseudospectral Method:

The pseudospectral method is a numerical technique used to approximate derivatives in a highly accurate way.

Unlike finite differences, which approximate derivatives using nearby points : local information, the pseudospectral method uses information from the entire grid of points : global information.
This makes the pseudospectral method very accurate for smooth problems, where the solution behaves nicely.

The pseudospectral method assumes that the function we want to differentiate can be expressed as a combination of simple mathematical functions like polynomials or trigonometric functions.

Using this representation, the method calculates derivatives exactly for these functions.

#### Steps:

- Choose grid points:  these are special points like Chebyshev or Gauss-Lobato points, which help reduce numerical errors.
    
- Construct a Derivative Matrix: This matrix, when multiplied by the function values at the grid points, gives the approximate derivatives at each point.
    
- Rescale for the Domain: If the problem's domain is not the standard interval [−1,1], we rescale the derivative matrices to match the actual domain size.
    
The `solve_ivp method` uses the explicit Runge-Kutta 4th/5th order method by default.

This method is adaptive and adjusts the time step dynamically based on local error estimates.

Adaptive time-stepping ensures numerical stability by reducing the time step when rapid changes are detected.

It avoids instabilities by reducing the time step when rapid changes are detected.

#### Advantages and Disadvantages of Pseudospectral Methods:

Advantages:

- Very accurate for smooth functions because it uses global information.
- Requires fewer points to achieve high accuracy compared to finite difference methods.

Disadvantages:

- Sensitive to discontinuities or sharp changes in the function (e.g., if the temperature changes very suddenly).
- Constructing and applying the derivative matrices can be computationally expensive for a large number of points.
- Requires careful treatment of boudary points.


#### `Scipy.integrate.solve.ivp` :

The `solve_ivp` function from the SciPy library is used to solve differential equations over time.

By default, `solve_ivp` uses the RK45 method, an explicit method that predicts the solution at the next time step by combining the current value and its derivatives, It dynamically adjusts the step size to ensure accuracy and avoid instability.

Adaptive time-stepping automatically adjusts the step size, balancing accuracy and efficiency.

Disadvantages:

- Explicit methods like RK45 can become inefficient for stiff problems (where some parts of the solution change much faster than others).


The pseudospectral method is like using a "big picture" view of the solution to compute derivatives very accurately, while solve_ivp is like a "smart driver" that knows when to slow down or speed up while solving the equation over time.

```python
class HeatEquationPseudospectral:
    """
    Solves the heat equation using pseudospectral derivatives and scipy.integrate.solve_ivp.
    
    Attributes:
    - spatial_grid (array-like): Spatial grid points for the solution.
    - initial_temperature (array-like): Initial temperature profile.
    - diffusivity (float): Thermal diffusivity (α).
    - frequency (float): Frequency of the oscillating surroundings source.
    - max_time (float): Maximum simulation time.
    - num_grid_points (int): Number of spatial grid points.
    - first_derivative_matrix (2D array-like): First derivative pseudospectral matrix.
    - second_derivative_matrix (2D array-like): Second derivative pseudospectral matrix.
    
    Methods:
    - get_time_derivative.
    - integrate_using_solve_ivp
    """
    
    def __init__(self, spatial_grid, initial_temperature, max_time, diffusivity=0.5, frequency=0.5):
        
        assert len(spatial_grid) == len(initial_temperature), "Spatial grid and initial temperature must have the same length."
        assert max_time > 0, "Maximum time must be positive."
        assert diffusivity > 0, "Diffusivity must be positive."
        assert frequency > 0, "Frequency must be positive."
    
        self.diffusivity = diffusivity
        self.frequency = frequency
        self.max_time = max_time
        self.spatial_grid = spatial_grid
        self.num_grid_points = len(spatial_grid)
        self.temperature = initial_temperature

        # Generate pseudospectral derivative matrices
        self.first_derivative_matrix = get_pseudospectral_first_derivative_matrix(self.num_grid_points - 1)
        self.second_derivative_matrix = get_pseudospectral_second_derivative_matrix(self.num_grid_points - 1)

        # Rescale matrices for the spatial interval [-L/2, L/2]
        domain_length = spatial_grid[-1] - spatial_grid[0]
        assert domain_length > 0, "Domain length must be positive."
        self.first_derivative_matrix *= (2 / domain_length)
        self.second_derivative_matrix *= (4 / domain_length**2)

    def get_time_derivative(self, current_time, temperature_profile):
        assert len(temperature_profile) == self.num_grid_points, "Temperature profile must match the number of grid points."
        
        surroundings = 20 + np.cos(self.frequency * current_time) * np.exp(-(self.spatial_grid / 1.5)**2)
        time_derivative = self.diffusivity * np.dot(self.second_derivative_matrix, temperature_profile) \
                          + (surroundings - temperature_profile)

        # Ensure boundary stability
        time_derivative[0] = 0
        time_derivative[-1] = 0

        assert len(time_derivative) == self.num_grid_points, "Time derivative must match the number of grid points."
        return time_derivative

    def integrate_using_solve_ivp(self, method="RK45", num_time_points=100):
        time_points = np.linspace(0, self.max_time, num_time_points)
        
        assert num_time_points > 1, "Number of time points must be greater than 1."
        assert isinstance(num_time_points, int), "Number of time points must be an integer."
        
        solution = solve_ivp(
            self.get_time_derivative,
            [0, self.max_time],
            self.temperature,
            t_eval=time_points,
            method=method
        )

        return solution.t, solution.y


# Define parameters
domain_length = 10
num_grid_points = 50
spatial_grid = np.linspace(-domain_length / 2, domain_length / 2, num_grid_points)
initial_temperature = 20 * (1 + np.exp(-spatial_grid**2))
maximum_time = 1.0

# Solve the heat equation using pseudospectral derivatives and solve_ivp
heat_equation_ps = HeatEquationPseudospectral(spatial_grid, initial_temperature, maximum_time)
time_points_ps, solutions_ps = heat_equation_ps.integrate_using_solve_ivp()

# Plot the solution
for i, current_time in enumerate(time_points_ps):
    if i % 10 == 0:
        plt.plot(spatial_grid, solutions_ps[:, i], label=f'time={current_time:.2f}')

plt.xlabel('x (spatial coordinate)')
plt.ylabel('T(x, t) (temperature profile)')
plt.title('Heat Equation Solved Using Pseudospectral Derivatives and solve_ivp')
plt.grid()
plt.show()
```

###### Why ODE Methods Are Not Suitable for Directly Solving This Problem:

The equation:

$\frac{\partial^2 T}{\partial x^2} - \gamma T^4 = 0 $

is an elliptic partial differential equation (PDE), not an ordinary differential equation (ODE). Here's why ODE methods are not suitable:

Using $ \Delta = B^2 - 4AC $ where :

- \(A = 1\) coefficient of ($ \frac{\partial^2 T}{\partial x^2}\$ )
- \(B = 0\) no mixed derivative term
- \(C = 0\) no second derivative in time

Substituting these values :

$ \Delta = 0^2 - 4(1)(0) = 0 $

This equation does not have any time-dependent terms $ (\frac{\partial T}{\partial t}) $ or $ (\frac{\partial^2 T}{\partial t^2}) $ . It involves only a spatial second derivative $ (\frac{\partial^2 T}{\partial x^2}) $ and a nonlinear term $ (-\gamma T^4)$ . Such equations are classified as **elliptic**.


This equation involves a second-order spatial derivative $ (\frac{\partial^2 T}{\partial x^2})$ , which cannot be directly handled by ODE solvers designed for time-dependent problems.

The term $ (-\gamma T^4) $ introduces nonlinearity, making analytical methods impractical. Instead, iterative numerical techniques are required.

ODE solvers like `scipy.integrate.solve_ivp` are designed for initial value problems , where the solution evolves with respect to one independent variable (usually time). However, this equation has boundary conditions instead of initial values.

To solve it, we reformulate the elliptic PDE into a "fake" parabolic PDE using the **relaxation method**.

##### <u>The Relaxation method <u> :
    
The **relaxation method** is a numerical technique used to iteratively solve steady-state problems, like elliptic PDEs. Here's how it works in simple terms:

1. Instead of solving the equation directly, we reformulate it as a time-dependent problem:
   $ \frac{\partial T}{\partial t} = \text{(original equation terms)} $ .
   Over time, the solution "relaxes" into the steady-state solution of the original equation.

2.  Start with an initial guess and repeatedly update the solution using a time-stepping method (e.g., Euler's method) until the solution stops changing significantly.

3.  The iterations are stopped when the solution converges, i.e., when  $ (\frac{\partial T}{\partial t} \approx 0) $.

Advantages: 
    
   - Handles nonlinear equations easily.
   - Does not require directly solving a complicated system of equations.

The system "relaxes" to the steady-state solution by simulating a pseudo-time evolution. The relaxation method allows us to solve this elliptic PDE by introducing artificial time evolution:

$\frac{\partial T}{\partial t} = \tau \left( \frac{\partial^2 T}{\partial x^2} - \gamma T^4 \right),$

where $ (\tau > 0) $ controls the rate of relaxation. We stop the iteration when the solution converges, i.e., when $ (\frac{\partial T}{\partial t} \approx 0) $ .

Relaxation method steps: 
    
1.Simplify the relaxation term by setting $ (\tau = 1)$ .
   
2.Use the nonlinear term $ (-\gamma T^4) $ as the source term instead of the surroundings function.

3.The boundary conditions $ (T(x = \pm L/2) = 20) $ are naturally enforced in the finite difference matrix.

4.Stop the iterations when the solution changes negligibly, i.e., when $ (\max(|\Delta T|)) $  is below a tolerance threshold.


In this problem, the relaxation method allows us to find the steady-state temperature distribution that satisfies the elliptic PDE. The artificial time evolution is used solely to reach the final solution.


```python
class RelaxationSolver:
    def __init__(self, spatial_grid, boundary_value, gamma, tau=1.0, dt=0.01):
        """
        Initialise the relaxation solver.
        
        Parameters:
        - boundary_value (float): Boundary temperature value.
        - gamma (float): Coefficient of the T^4 term.
        - tau (float): Relaxation parameter (default=1.0).
        - dt (float): Time step for relaxation (default=0.01).
        
        Methods:
        - get_time_drivative
        - solve : solves the elliptic PDE using the relaxation method.
        """
        self.spatial_grid = spatial_grid
        self.boundary_value = boundary_value
        self.gamma = gamma
        self.tau = tau
        self.dt = dt
        self.temperature = np.full_like(spatial_grid, boundary_value)  # Initial guess
        
        # Finite difference matrix for second derivative
        N = len(spatial_grid)
        dx = spatial_grid[1] - spatial_grid[0]
        self.D2 = np.zeros((N, N))
        for i in range(1, N - 1):
            self.D2[i, i - 1] = 1.0
            self.D2[i, i] = -2.0
            self.D2[i, i + 1] = 1.0
        self.D2 /= dx**2
    
    def get_time_derivative(self):
        
        """
        Compute the time derivative for the relaxation method.
        
        """
        T = self.temperature
        time_derivative = self.tau * (np.dot(self.D2, T) - self.gamma * T**4)
        
        # Enforce boundary conditions
        time_derivative[0] = 0
        time_derivative[-1] = 0
        return time_derivative
    
    def solve(self, max_iterations=10000, tolerance=1e-6):
        
        """
        Solve the elliptic PDE using the relaxation method.

        """
        for iteration in range(max_iterations):
            time_derivative = self.get_time_derivative()
            self.temperature += self.dt * time_derivative
            
            # Check convergence
            if np.max(np.abs(time_derivative)) < tolerance:
                print(f"Converged in {iteration} iterations.")
                break
        else:
            print("Did not converge within the maximum number of iterations.")
        
        return self.temperature

# Define parameters
L = 10  # Length of the domain
N = 50  # Number of grid points
spatial_grid = np.linspace(-L / 2, L / 2, N)
boundary_value = 20
gamma = 0.0001

# Solve the elliptic PDE using relaxation
solver = RelaxationSolver(spatial_grid, boundary_value, gamma)
solution = solver.solve()

# Plot the solution
plt.figure(figsize=(10, 6))
plt.plot(spatial_grid, solution, label="T(x)",color="b")
plt.axhline(boundary_value, color="r", linestyle="--", label="Boundary Value (T=20)")
plt.xlabel("x (spatial coordinate)")
plt.ylabel("T(x) (temperature)")
plt.title("Solution to Elliptic PDE Using Relaxation")
plt.grid()
plt.legend()
plt.show()
```

#### Interpretation:

The temperature profile starts at $ T(x = −L/2)=20 $ and ends at $ T(x=L/2)=20 $ as dictated by the boundary conditions.

The relaxation method converged after 3708 iterations, which indicates that the iterative updates successfully found a stable solution satisfying both the differential equation and boundary conditions.

The plot shows a parabolic shape for $T(x)$, starting and ending at $20°C$ at the boundaries and dipping to a minimum in the middle.

We can think of it in this way, Heat diffuses from hotter regions (the center of the bar) to colder regions (the boundaries) by the process of diffusion. The $T^4$ nonlinearity causes the profile to flatten out and stabilise, ensuring the solution does not diverge or oscillate.
